```python
import pandas as pd
from tqdm import tqdm
import os
import ast
```

```python
with open('../cl_seqs_filtered.txt', 'r') as f:
    seqs_to_keep = f.readlines()

seqs_to_keep = [s.strip() for s in seqs_to_keep]
```

```python
df_list = []
for item_name in tqdm(os.listdir("/global/scratch/users/claireleblanc/transcription_factors/")):
    try: 
        preds = pd.read_csv(f"/global/scratch/users/claireleblanc/transcription_factors/{item_name}/{item_name}_orthologs_consensus_all_six.csv", index_col=0)
        df_list.append(preds[[n in seqs_to_keep for n in preds['name']]])
    except: 
        print(item_name)
```

```python
filtered_seqs = pd.concat(df_list, ignore_index = True)
filtered_seqs
```

```python
# Currently just missing sequences from Flo
len(seqs_to_keep)
```

```python
len(set(seqs_to_keep).difference(filtered_seqs['name']))
```

```python
len(set(filtered_seqs['name']).difference(seqs_to_keep))
```

```python
filtered_seqs['length'] = filtered_seqs['sequence'].str.len()
```

```python
filtered_seqs[filtered_seqs['length'] <= 53]
```

```python
# Write to output file
filtered_seqs.to_csv('../predictions/combined_preds/y1000_filtered_all_six_combined.csv')
```
