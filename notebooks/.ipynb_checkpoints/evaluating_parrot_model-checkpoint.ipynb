```python
# Predictions for things in the test set
predictions = pd.read_table('../models/consensus_model_all_six_predictions.tsv', header=None, sep=' ', names=list(range(1444)))
predictions['label'] = [s.split('_')[-1] for s in predictions[0]]
predictions
```

```python
true_df = predictions[predictions['label'] == 'TRUE']
true_vals_ls = true_df.iloc[:, 2:1443].apply(lambda row: list(row.dropna()), axis=1).tolist()
```

```python
true_vals_ls = [a for ls in true_vals_ls for a in ls]
```

```python
pred_df = predictions[predictions['label'] == 'PRED']
pred_vals_ls = pred_df.iloc[:, 2:1443].apply(lambda row: list(row.dropna()), axis=1).tolist()
pred_vals_ls = [a for ls in pred_vals_ls for a in ls]
```

```python
len(true_vals_ls)
```

```python
len(pred_vals_ls)
```

```python
sns.boxplot(x=true_vals_ls, y=pred_vals_ls,
           flierprops={"marker": ".", "markersize": 3, "color": "grey"})
sns.despine()
pearson_corr, pearson_p = pearsonr(true_vals_ls, pred_vals_ls)
spearman_corr, spearman_p = spearmanr(true_vals_ls, pred_vals_ls)
plt.title(f"Performance of Parrot Consensus NN on test set\nPearson corr={pearson_corr:.3}\nSpearman corr={spearman_corr:.3}")

plt.xticks([0, 1, 2, 3, 4, 5, 6], ["0", "1/6", "2/6", "3/6", "4/6", "5/6", "6/6"])
plt.xlabel("Number of predictors predicting AD")
plt.ylabel("Predicted number of predictors predicting AD")
```

```python
def round_to_interval(value, interval):
    return round(value / interval) * interval

rounded_pred_ls = [round_to_interval(n, 1/6) for n in pred_vals_ls]
```

```python
labels = [str(round(v, 8)) for v in [0, 1/6, 2/6, 3/6, 4/6, 5/6, 1]]

cm = confusion_matrix(
    [str(round(v, 8)) for v in true_vals_ls],
    [str(round(v, 8)) for v in rounded_pred_ls],
    labels=labels  # ensures correct ordering
)

ConfusionMatrixDisplay(cm, display_labels=labels).plot()
```
