# PADDLE test set proteins (good for creating hits list)

```python
# Figure 3 source data 
paddle_test_set = pd.read_excel('Paddle_test_set_sequences.xlsx', sheet_name='PADDLE test set predictions')

# Figure 1 source data (1 and 2)
paddle_ads = pd.read_excel("list_of_ADs.xlsx")
paddle_full_seqs = pd.read_excel('screen_data.xlsx', sheet_name='TF sequences, see Fig 1')
```

```python
# These are full length TFs not used to train PADDLE
paddle_test_set['protein'].value_counts()
```

```python
paddle_test_set[paddle_test_set['protein'] == 'MET31']
```

```python
test_TFs = paddle_test_set['protein'].dropna().unique()
```

```python
paddle_ads_subset = paddle_ads[[n in test_TFs for n in paddle_ads['protein']]]
```

```python
paddle_ADs_w_full_seqs = paddle_ads_subset.merge(paddle_full_seqs)
```

```python
# Check that AD is in the full sequence
for i in paddle_ADs_w_full_seqs.index: 
    print(paddle_ADs_w_full_seqs.loc[i, 'sequence'] in paddle_ADs_w_full_seqs.loc[i, 'protein sequence'])
```

```python
paddle_ADs_w_full_seqs_filtered = paddle_ADs_w_full_seqs.drop_duplicates(['protein', 'protein sequence'])
```

```python
# Write to fasta file
with open('paddle_test_seqs.fasta', 'w+') as f: 
    for i in paddle_ADs_w_full_seqs_filtered.index:
        f.write(">" + paddle_ADs_w_full_seqs_filtered.loc[i, 'protein'] + "\n")
        f.write(paddle_ADs_w_full_seqs_filtered.loc[i, 'protein sequence'] + "\n")
```

```python
# Write to file for parrot
with open('paddle_test_seqs.tsv', 'w+') as f: 
    for i in paddle_ADs_w_full_seqs_filtered.index:
        f.write(paddle_ADs_w_full_seqs_filtered.loc[i, 'protein'] + " ")
        f.write(paddle_ADs_w_full_seqs_filtered.loc[i, 'protein sequence'] + "\n")
```

```python
paddle_ADs_w_full_seqs.to_csv('paddle_test_seqs.csv')
```
