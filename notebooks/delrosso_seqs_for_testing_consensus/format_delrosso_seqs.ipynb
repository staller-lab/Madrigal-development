```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def read_fasta(file):
    with open(file, 'r') as f:
        lines = f.readlines()
    names = []
    seqs = []
    seq = ""
    for line in lines: 
        if line.startswith('>'): 
            names.append(line.replace(">", "").strip())
            if len(names) > 1:
                seqs.append(seq)
                seq = ""
        else:
            seq += line.strip()
    
    seqs.append(seq)

    return pd.DataFrame({'name' : names, 'sequence' : seqs})
```

```python
TFs_for_test_set = ['THAP12', 'CEBPG', 'FOXJ2', 'HOXA2', 'NR3C2', 'NR2C1', 'PKNOX2',
       'THRA', 'ZBTB20', 'MAFG', 'MBD6', 'ELF3', 'TEAD1', 'ZSCAN23',
       'ZNF121', 'SOX1', 'ZNF532', 'CREB3', 'ZNF683', 'ZNF80', 'NFE2L3',
       'NFATC2', 'L3MBTL3', 'ATOH8', 'NR2F1']
```

```python
ad_data = pd.read_excel('List_of_delrosso_ADs.xlsx')
all_data = pd.read_excel('delrosso_data.xlsx')
```

```python
all_data_filtered = all_data[[s in TFs_for_test_set for s in all_data['HGNC Symbol']]] #["HGNC Symbol"].value_counts()
```

```python
all_data_filtered[all_data_filtered['HGNC Symbol'] == 'CREB3']
```

```python
ad_data_filtered = ad_data[[s in TFs_for_test_set for s in ad_data['HGNC symbol']]]
ad_data_filtered
```

```python
# These are the full sequences of the TFs of interest
lambert_TFs = read_fasta('lambert_TFs_10-21-24_with_DBD_coords.fasta')
lambert_TFs["UniProt ID"] = lambert_TFs["name"].str.split("|").str[1]
lambert_TFs = lambert_TFs.rename(columns={'sequence' : 'full_seq'})
```

```python
# We need a way to translate the uniprot IDs to get out the TFs of interest
name_df = pd.read_table('lambert_tfs_gene_names.tsv')
uniprot_IDs_for_test_set = name_df[[n in TFs_for_test_set for n in name_df['Gene Names (primary)']]]['Entry'].values

uniprot_TFs_for_test_set = lambert_TFs[[n in uniprot_IDs_for_test_set for n in lambert_TFs['UniProt ID']]]
```

```python
# Add ADs to test TFs that have ADs
test_TFs_w_ADs = uniprot_TFs_for_test_set.merge(ad_data_filtered, how='outer')
```

```python
test_TFs_w_ADs.to_csv('delrosso_test_TFs_w_ADs.csv')
```

```python
test_TFs_w_ADs_filtered = test_TFs_w_ADs.drop_duplicates(['UniProt ID', 'full_seq'])
```

```python
with open('delrosso_test_TFs_w_ADs.fasta', 'w+') as f:
    for i in test_TFs_w_ADs_filtered.index:
        f.write(">" + test_TFs_w_ADs_filtered.loc[i, 'UniProt ID'] + "\n")
        f.write(test_TFs_w_ADs_filtered.loc[i, 'full_seq'] + "\n")
```

```python
with open('delrosso_test_TFs_w_ADs.tsv', 'w+') as f:
    for i in test_TFs_w_ADs_filtered.index:
        f.write(test_TFs_w_ADs_filtered.loc[i, 'UniProt ID'] + " ")
        f.write(test_TFs_w_ADs_filtered.loc[i, 'full_seq'] + "\n")
```
