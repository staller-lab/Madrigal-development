```python
import pandas as pd
import protfasta
```

```python
train_seqs_dict = protfasta.read_fasta("data/train_seqs_with_test_overlap_removed.fasta")
train_seqs_df = pd.DataFrame({"id" : train_seqs_dict.keys(), "seq" : train_seqs_dict.values()})
train_seqs_df
```

```python
# Can remove claire's sequences because predictors already ran
```

```python
cl_seqs = protfasta.read_fasta("data/cl_seqs.fasta", invalid_sequence_action='remove')
cl_seqs_df = pd.DataFrame({"id": cl_seqs.keys(), "seq": cl_seqs.values()})
cl_seqs_df
```

```python
non_cl_train_seqs = train_seqs_df[~train_seqs_df["seq"].isin(cl_seqs_df["seq"])]
non_cl_train_seqs
```

```python
protfasta.write_fasta(dict(zip(non_cl_train_seqs["id"], non_cl_train_seqs["seq"])), "data/train_seqs_with_test_overlap_removed_sk_only.fasta")
```

```python
# ! mkdir ../data/non_cl_train_seqs_split
```

```python
# df = your loaded dataframe with columns: id, seq
chunk_size = 1000
file_paths = []

for i in range(0, len(non_cl_train_seqs), chunk_size):
    chunk = non_cl_train_seqs.iloc[i:i + chunk_size]
    fasta_content = ""
    for _, row in chunk.iterrows():
        fasta_content += f">{row['id']}\n{row['seq']}\n"

    file_path = f"data/splits/non_cl_train_seqs_split/pt{i//chunk_size + 1}.fasta"
    with open(file_path, "w") as f:
        f.write(fasta_content)
    file_paths.append(file_path)
```
